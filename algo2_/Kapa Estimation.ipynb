```python
import pandas as pd

def set_datetime_index(
    df, 
    datetime_col="time", 
    tz=None, 
    drop_cols=None,
    drop_nat=True
):
    """
    Convert a datetime column to index for a single dataset,
    drop optional columns, and sort by index date (microseconds preserved).
    """
    df = df.copy()
    
    # Convert to datetime (mixed format, preserve microseconds if available)
    df[datetime_col] = pd.to_datetime(df[datetime_col], errors="coerce", utc=False)
    
    # If timezone conversion is requested
    if tz:
        # If datetimes are naive → localize, otherwise convert
        if df[datetime_col].dt.tz is None:
            df[datetime_col] = df[datetime_col].dt.tz_localize(tz)
        else:
            df[datetime_col] = df[datetime_col].dt.tz_convert(tz)
    
    # Drop invalid rows (NaT) if requested
    if drop_nat:
        df = df.dropna(subset=[datetime_col])
    
    # Drop unwanted columns
    if drop_cols:
        df = df.drop(columns=[c for c in drop_cols if c in df.columns], errors="ignore")
    
    # Set datetime as index and sort ascending
    df = df.set_index(datetime_col).sort_index()
    
    return df
```

```python
def remove_duplicates(df, subset=None, keep="first"):
    """
    Remove duplicate rows from a DataFrame.

    Parameters
    ----------
    df : pd.DataFrame
        Input dataset (with datetime index or not).
    subset : list, optional
        Columns to consider when identifying duplicates.
        If None, considers all columns.
    keep : {"first", "last", False}, default "first"
        - "first": keep the first occurrence
        - "last": keep the last occurrence
        - False: drop all duplicates

    Returns
    -------
    pd.DataFrame
        DataFrame without duplicate rows.
    """
    return df.drop_duplicates(subset=subset, keep=keep)
```

# Filter base on minute

```python
import pandas as pd

def split_by_minutes(df, freq_minutes=10):
    """
    Split a time-indexed DataFrame into consecutive minute-based windows
    and return a dict with proper interval names.

    Parameters
    ----------
    df : pd.DataFrame
        Input dataset with a DateTimeIndex.
    freq_minutes : int, default=10
        Length of each time window in minutes.

    Returns
    -------
    dict
        Keys are interval labels like "minute_0_to_10", "minute_10_to_20", ...
        Values are DataFrames filtered to that interval.
    """
    if not isinstance(df.index, pd.DatetimeIndex):
        raise ValueError("DataFrame index must be a DateTimeIndex")

    # Ensure sorted by time
    df = df.sort_index()

    # Calculate elapsed minutes from the start
    start_time = df.index[0]
    elapsed_minutes = ((df.index - start_time).total_seconds() // 60).astype(int)

    # Integer group ID based on freq
    group_ids = (elapsed_minutes // freq_minutes).astype(int)

    result = {}
    for group_num, group_df in df.groupby(group_ids):
        start_m = group_num * freq_minutes
        end_m = start_m + freq_minutes
        key = f"minute_{start_m}_to_{end_m}"
        result[key] = group_df

    return result
```

```python
import numpy as np
import pandas as pd
from scipy.optimize import curve_fit

def interpolate_mid_price(orderbook_df, start_time, end_time, timestamps):
    """
    Interpolate the mid price between the first (t_0) and last (t_4) order book snapshot.
    
    Parameters:
    - orderbook_df: DataFrame containing the order book snapshots at t_0 and t_4.
    - start_time: The timestamp of the first order book snapshot (t_0).
    - end_time: The timestamp of the last order book snapshot (t_4).
    - timestamps: List of timestamps at which to estimate the mid price.
    
    Returns:
    - A pandas Series with interpolated mid prices for each timestamp.
    """
    # Extract the best bid and ask prices from t_0 and t_4
    bid_t0 = orderbook_df.loc[start_time, 'bid_price_1']
    ask_t0 = orderbook_df.loc[start_time, 'ask_price_1']
    bid_t4 = orderbook_df.loc[end_time, 'bid_price_1']
    ask_t4 = orderbook_df.loc[end_time, 'ask_price_1']
    
    # Linear interpolation of mid prices (can use any interpolation method)
    def interpolate(t):
        alpha = (t - start_time) / (end_time - start_time)
        mid_price = (1 - alpha) * ((bid_t0 + ask_t0) / 2) + alpha * ((bid_t4 + ask_t4) / 2)
        return mid_price

    return pd.Series([interpolate(t) for t in timestamps], index=timestamps)

def align_trades_to_quotes(trades_df, quotes_df, tolerance="1s"):
    """
    Align trades to the closest order book quote before the trade timestamp.
    
    Parameters:
    - trades_df: DataFrame containing trades with timestamps and prices.
    - quotes_df: DataFrame containing order book snapshots (quotes) with timestamps.
    - tolerance: Maximum allowed gap between trade and quote timestamps.
    
    Returns:
    - DataFrame with trades aligned to their closest quotes.
    """
    # Ensure both DataFrames are sorted by timestamp
    trades_df = trades_df.sort_index()
    quotes_df = quotes_df.sort_index()

    # Align trades with the closest quote before each trade timestamp
    aligned_trades = pd.merge_asof(
        trades_df, quotes_df, left_index=True, right_index=True,
        direction="backward",  # Align to the previous quote before each trade
        tolerance=pd.to_timedelta(tolerance),  # Tolerance window for the gap
        allow_exact_matches=True
    )

    return aligned_trades

def _levels_and_lambdas(trades_aligned, tick_size=None, volume_col="volume"):
    """
    Given trades with columns ['price','quote_px', volume_col], compute:
    - level: continuous |p_trade - quote|  OR  integer tick distance floor(|...|/tick_size)
    - lambda(level): sum of volumes per level
    """
    diff = (trades_aligned["price"] - trades_aligned["quote_px"]).abs()
    if tick_size is not None and tick_size > 0:
        levels = np.floor(diff / tick_size).astype(int)
    else:
        levels = diff  # continuous levels (price units)

    # Aggregate volume by level
    grp = trades_aligned.groupby(levels)[volume_col].sum()
    # Remove zero or negative (just in case)
    grp = grp[grp > 0]

    x = grp.index.to_numpy(dtype=float)   # levels
    y = grp.to_numpy(dtype=float)         # lambdas (volumes)
    return x, y

def _fit_exp(x, y):
    """
    Fit y ≈ α * exp(-κ x). Returns (alpha, kappa, diagnostics)
    Robust to few points; falls back to log-linear OLS if curve_fit fails.
    """
    eps = 1e-10
    y_adj = np.maximum(y, eps)

    # Initial guesses
    a0 = float(y_adj.max())
    # crude slope init: -k ≈ slope of log y vs x
    if len(x) >= 2:
        slope_init = - (np.log(y_adj[-1]) - np.log(y_adj[0])) / (x[-1] - x[0] + eps)
        b0 = max(slope_init, eps)
    else:
        b0 = 1.0

    def model(t, a, b):
        return a * np.exp(-b * t)

    try:
        (a_hat, b_hat), _ = curve_fit(
            model, x, y_adj,
            p0=(a0, b0),
            bounds=([0.0, 0.0], [np.inf, np.inf]),
            method="dogbox",
            maxfev=10000,
        )
        fitted = model(x, a_hat, b_hat)
    except Exception:
        # Fallback: log-linear OLS on log y = log a - k x
        A = np.vstack([np.ones_like(x), -x]).T
        coef, *_ = np.linalg.lstsq(A, np.log(y_adj), rcond=None)
        log_a, kappa = coef[0], coef[1]
        a_hat = float(np.exp(log_a))
        b_hat = float(kappa)
        fitted = a_hat * np.exp(-b_hat * x)

    # Simple R^2 for diagnostics (on log scale to reflect exponential fit)
    ss_res = np.sum((np.log(y_adj) - np.log(np.maximum(fitted, eps)))**2)
    ss_tot = np.sum((np.log(y_adj) - np.log(y_adj).mean())**2) if len(y_adj) > 1 else np.nan
    r2_log = 1 - ss_res/ss_tot if ss_tot and ss_tot > 0 else np.nan

    return float(a_hat), float(b_hat), {"n_points": len(x), "r2_log": r2_log}

def estimate_intensity_once(
    trades_df,
    orderbook_df,
    price_type="mid",
    window=None,      # e.g. '30s', '2min', or None for "all data"
    end_time=None,    # if window is set, this is the right edge; default = last trade time
    tick_size=None,   # set if you want integer 'levels' in ticks
    side=None,        # 'buy', 'sell', or None for both
    volume_col="volume",
    tolerance="1s"    # max allowed gap from trade to last quote
):
    """
    Estimate (alpha, kappa) once over a single time slice.
    - trades_df: index = time_Iran (tz-aware), columns: ['symbol','price','volume','type']
    - orderbook_df: index = time_Iran (tz-aware), columns include ['bid_price_1','ask_price_1']
    """
    # Optional side filter (be sure 'type' is 'buy'/'sell' semantics for aggressive side)
    td = trades_df.copy()
    if side in ("buy", "sell"):
        td = td[td["type"] == side]

    # Cut to time window
    td = td.sort_index()
    if window is not None:
        if end_time is None:
            end_time = td.index.max()
        start_time = pd.Timestamp(end_time) - pd.to_timedelta(window)
        td = td.loc[(td.index > start_time) & (td.index <= end_time)]
        if td.empty:
            return np.nan, np.nan, {"n_points": 0, "r2_log": np.nan}

    # Build quotes and align trades to latest quote <= trade time
    q = _make_quotes(orderbook_df, price_type=price_type)
    q = q.sort_index()
    td = td.sort_index()
    aligned = pd.merge_asof(
        td, q, left_index=True, right_index=True,
        direction="backward",
        tolerance=pd.to_timedelta(tolerance),
        allow_exact_matches=True,
    ).dropna(subset=["quote_px", "price", volume_col])

    if aligned.empty:
        return np.nan, np.nan, {"n_points": 0, "r2_log": np.nan}

    # Build levels and empirical lambdas
    x, y = _levels_and_lambdas(aligned, tick_size=tick_size, volume_col=volume_col)
    if len(x) < 2:
        return np.nan, np.nan, {"n_points": len(x), "r2_log": np.nan}

    # Fit exponential
    alpha, kappa, diag = _fit_exp(x, y)
    return alpha, kappa, diag

def estimate_intensity_series(
    trades_df,
    orderbook_df,
    price_type="mid",
    window="30s",
    step="1s",
    tick_size=None,
    side=None,
    volume_col="volume",
    tolerance="1s"
):
    """
    Rolling time series of (alpha, kappa) over the dataset.
    Each row corresponds to the window (t - window, t], stepped by 'step'.
    """
    td = trades_df.sort_index()
    if td.empty:
        return pd.DataFrame(columns=["alpha", "kappa", "r2_log"])

    end_times = pd.date_range(td.index.min() + pd.to_timedelta(window),
                              td.index.max(), freq=step)
    out = []
    for t in end_times:
        a, k, diag = estimate_intensity_once(
            td, orderbook_df,
            price_type=price_type,
            window=window,
            end_time=t,
            tick_size=tick_size,
            side=side,
            volume_col=volume_col,
            tolerance=tolerance
        )
        out.append((t, a, k, diag.get("r2_log", np.nan)))

    res = pd.DataFrame(out, columns=["time", "alpha", "kappa", "r2_log"]).set_index("time")
    return res
```

# Last_qoute

```python
import numpy as np
import pandas as pd
from scipy.optimize import curve_fit
from tqdm.auto import tqdm
import matplotlib.pyplot as plt


# -----------------------------
# Midprice calculation
# -----------------------------
def get_mid_price(orderbook_row):
    """
    Calculate mid price for an order book row using best bid and ask prices.
    """
    bid_price = orderbook_row['bid_price_1']
    ask_price = orderbook_row['ask_price_1']
    if pd.isna(bid_price) or pd.isna(ask_price):
        return np.nan
    mid_price = (bid_price + ask_price) / 2
    return mid_price


# -----------------------------
# Align trades with quotes
# -----------------------------
def align_trades_to_last_quote(trades_df, orderbook_df, tolerance="1s"):
    """
    Align trades to the closest order book snapshot before the trade timestamp.
    """
    trades_df = trades_df.sort_index()
    orderbook_df = orderbook_df.sort_index()

    orderbook_df = orderbook_df.ffill()  # forward fill missing quotes

    aligned_trades = pd.merge_asof(
        trades_df, orderbook_df,
        left_index=True, right_index=True,
        direction="backward",
        tolerance=pd.to_timedelta(tolerance),
        allow_exact_matches=True
    )

    aligned_trades['midprice'] = aligned_trades.apply(get_mid_price, axis=1)
    return aligned_trades


# -----------------------------
# Aggregate into levels
# -----------------------------
def _levels_and_lambdas(trades_aligned, tick_size=None, volume_col="volume"):
    """
    Compute trade intensities λ(ℓ) by level.
    """
    diff = (trades_aligned["price"] - trades_aligned["midprice"]).abs()
    if tick_size is not None and tick_size > 0:
        levels = np.floor(diff / tick_size).astype(int)
    else:
        levels = diff

    grp = trades_aligned.groupby(levels)[volume_col].sum()
    grp = grp[grp > 0]

    x = grp.index.to_numpy(dtype=float)   # levels
    y = grp.to_numpy(dtype=float)         # lambdas (volumes)
    return x, y


# -----------------------------
# Fit exponential λ(ℓ) = α exp(-κℓ)
# -----------------------------
def _fit_exp(x, y):
    """
    Fit exponential curve and return (alpha, kappa, fitted_values, diagnostics).
    """
    eps = 1e-10
    y_adj = np.maximum(y, eps)

    a0 = float(y_adj.max())
    if len(x) >= 2:
        slope_init = - (np.log(y_adj[-1]) - np.log(y_adj[0])) / (x[-1] - x[0] + eps)
        b0 = max(slope_init, eps)
    else:
        b0 = 1.0

    def model(t, a, b):
        return a * np.exp(-b * t)

    try:
        (a_hat, b_hat), _ = curve_fit(
            model, x, y_adj,
            p0=(a0, b0),
            bounds=([0.0, 0.0], [np.inf, np.inf]),
            method="dogbox",
            maxfev=10000,
        )
        fitted = model(x, a_hat, b_hat)
    except Exception:
        # fallback log-linear OLS
        A = np.vstack([np.ones_like(x), -x]).T
        coef, *_ = np.linalg.lstsq(A, np.log(y_adj), rcond=None)
        log_a, kappa = coef[0], coef[1]
        a_hat = float(np.exp(log_a))
        b_hat = float(kappa)
        fitted = a_hat * np.exp(-b_hat * x)

    ss_res = np.sum((np.log(y_adj) - np.log(np.maximum(fitted, eps)))**2)
    ss_tot = np.sum((np.log(y_adj) - np.log(y_adj).mean())**2) if len(y_adj) > 1 else np.nan
    r2_log = 1 - ss_res/ss_tot if ss_tot and ss_tot > 0 else np.nan

    return float(a_hat), float(b_hat), fitted, {"n_points": len(x), "r2_log": r2_log}


# -----------------------------
# Single-window estimation
# -----------------------------
def estimate_intensity_once(
    trades_df, orderbook_df,
    price_type="mid", window=None, end_time=None,
    tick_size=None, side=None, volume_col="volume", tolerance="100s"
):
    """
    Estimate intensity parameters for one time window.
    Returns alpha, kappa, x, y_true, y_hat, diag
    """
    td = trades_df.copy()
    if side in ("buy", "sell"):
        td = td[td["type"] == side]

    td = td.sort_index()
    if window is not None:
        if end_time is None:
            end_time = td.index.max()
        start_time = pd.Timestamp(end_time) - pd.to_timedelta(window)
        td = td.loc[(td.index > start_time) & (td.index <= end_time)]
        if td.empty:
            return np.nan, np.nan, [], [], [], {"n_points": 0, "r2_log": np.nan}

    aligned_trades = align_trades_to_last_quote(td, orderbook_df)
    if aligned_trades.empty:
        return np.nan, np.nan, [], [], [], {"n_points": 0, "r2_log": np.nan}

    x, y = _levels_and_lambdas(aligned_trades, tick_size=tick_size, volume_col=volume_col)
    if len(x) < 2:
        return np.nan, np.nan, x, y, [], {"n_points": len(x), "r2_log": np.nan}

    alpha, kappa, y_hat, diag = _fit_exp(x, y)
    return alpha, kappa, x, y, y_hat, diag


# -----------------------------
# Rolling series estimation
# -----------------------------
def estimate_intensity_series(
    trades_df, orderbook_df,
    price_type="mid", window="600s", step="1s",
    tick_size=None, side=None, volume_col="volume", tolerance="100s"
):
    """
    Rolling time series of (alpha, kappa).
    """
    td = trades_df.sort_index()
    if td.empty:
        return pd.DataFrame(columns=["alpha", "kappa", "r2_log"])

    end_times = pd.date_range(td.index.min() + pd.to_timedelta(window),
                              td.index.max(), freq=step)
    
    out = []
    for t in tqdm(end_times, desc="Calculating Intensity Parameters", unit="window"):
        a, k, x, y, y_hat, diag = estimate_intensity_once(
            td, orderbook_df,
            price_type=price_type,
            window=window, end_time=t,
            tick_size=tick_size,
            side=side,
            volume_col=volume_col,
            tolerance=tolerance
        )
        out.append((t, a, k,x, y, y_hat, diag.get("r2_log", np.nan)))
    res = pd.DataFrame(out, columns=["time", "alpha", "kappa",'x','y','y_hat' ,"r2_log"]).set_index("time")
    return res

# -----------------------------
# Plotting utility
# -----------------------------
def plot_fit(x, y, y_hat, alpha, kappa, title="Lambda Fit"):
    """
    Plot empirical vs fitted lambdas.
    """
    plt.figure(figsize=(7,5))
    plt.scatter(x, y, color="blue", label="True λ(level)")
    plt.plot(x, y_hat, color="red", label=f"Fitted λ(ℓ) = {alpha:.2f} exp(-{kappa:.2f}ℓ)")
    plt.xlabel("Level ℓ")
    plt.ylabel("λ(ℓ)")
    plt.title(title)
    plt.legend()
    plt.grid(True, linestyle="--", alpha=0.6)
    plt.show()
```
