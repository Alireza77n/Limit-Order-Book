```python

def preprocess_lob(df: pd.DataFrame, keep_col: str = "time") -> pd.DataFrame:
    """
    Preprocess LOB dataset: keep only one time column, remove other time columns.
    
    Args:
        df (pd.DataFrame): raw dataset
        keep_col (str): which time column to keep (default: 'time')

    Returns:
        pd.DataFrame: cleaned dataset
    """
    # Identify time-like columns
    time_cols = ["time", "time_Iran", "request_time", "response_time"]

    # Drop all except keep_col
    drop_cols = [c for c in time_cols if c in df.columns and c != keep_col]
    df = df.drop(columns=drop_cols, errors="ignore")

    # Parse datetime
    df[keep_col] = pd.to_datetime(df[keep_col], errors="coerce")

    # Drop rows with missing timestamps
    df = df.dropna(subset=[keep_col])

    return df
```

```python
import pandas as pd
import numpy as np

class LOBVolatilityEstimator:
    def __init__(self, 
                 time_col: str = "time", 
                 price_col: str = "last_trade_price",
                 method: str = "rms",   # "rms", "std", "log-realized"
                 min_points: int = 10):
        """
        Flexible volatility estimator for Limit Order Book datasets.

        Args:
            time_col (str): column containing timestamps
            price_col (str): column for price (last_trade_price or mid-price)
            method (str): 
                - "rms" = sqrt(mean of squared raw increments Δs)
                - "std" = standard deviation of raw increments Δs
                - "log-realized" = realized volatility via sqrt(sum of squared log-returns)
            min_points (int): minimum ticks required to compute volatility
        """
        self.time_col = time_col
        self.price_col = price_col
        self.method = method
        self.min_points = min_points

    def _get_price_series(self, df: pd.DataFrame) -> pd.Series:
        """Extract price series (last trade or mid-price)."""
        if self.price_col and self.price_col in df.columns:
            return df[self.price_col].astype(float)
        if "bid_price_1" in df.columns and "ask_price_1" in df.columns:
            return ((df["bid_price_1"] + df["ask_price_1"]) / 2).astype(float)
        raise ValueError("No valid price column found!")

    def _calc_volatility(self, prices: pd.Series) -> float:
        """Calculate volatility for one segment of prices."""
        if len(prices) < self.min_points:
            return np.nan

        vals = prices.values.astype(float)

        if self.method == "rms":  # root mean square of raw increments
            diffs = np.diff(vals)
            print(prices)
            return np.sqrt(np.mean(diffs**2))

        elif self.method == "std":  # sample std of raw increments
            diffs = np.diff(vals)
            return np.std(diffs, ddof=1)

        elif self.method == "log-realized":  # realized volatility from log returns
            log_ret = np.diff(np.log(vals))
            return np.sqrt(np.sum(log_ret**2))

        else:
            raise ValueError("method must be 'rms', 'std', or 'log-realized'")

    def run(self, df: pd.DataFrame, window: str = "1D",
            start: str = None, end: str = None) -> pd.DataFrame:
        """
        Estimate volatility per time window.

        Args:
            df (pd.DataFrame): input dataset
            window (str): resample window (e.g. '1D', '6H', '30T')
            start (str): optional custom start time
            end (str): optional custom end time

        Returns:
            pd.DataFrame with volatility per window
        """
        df = df.copy()
        df[self.time_col] = pd.to_datetime(df[self.time_col])

        # Restrict time range if requested
        if start:
            df = df[df[self.time_col] >= pd.to_datetime(start)]
        if end:
            df = df[df[self.time_col] <= pd.to_datetime(end)]

        df = df.set_index(self.time_col).sort_index()
        price_series = self._get_price_series(df)

        results = []
        for win_start, group in price_series.resample(window):
            vol = self._calc_volatility(group.dropna())
            results.append({
                "window_start": win_start,
                "window_end": win_start + pd.Timedelta(window),
                "n_ticks": len(group),
                "vol": vol
            })
        return pd.DataFrame(results)
```
